```python
# tensor_order = 3

# A' = Q_ik Q_jl A_kl

indices = [1, 2, 3]

value = 0

a = [[45, 12, 30],[12, 15, 0],[30, 0, -5]]

q = [[0.888, 0.2284, 0.4], [0.14221, -0.96176, 0.23408], [-0.43803, 0.15092, 0.88620]]

tr_tensor = [[], [], []]

value = 0

for item, i in enumerate(indices):
    for item, j in enumerate(indices):
        for item, k in enumerate(indices):
            for item, l in enumerate(indices):
               value += q[i-1][k-1]*q[j-1][l-1]*a[k-1][l-1]
        tr_tensor[i-1].insert(j-1, round(value, 2))
        value = 0

print(tr_tensor[0])
print(tr_tensor[1])
print(tr_tensor[2])
```
